```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cars_raw.csv")
print(df.head()) 
print("\
Data types of columns:")
print(df.dtypes)
```

```python

# Read the CSV file
df = pd.read_csv('cars_raw.csv')

# Clean price column by removing '$' and ',' and handling 'Not Priced' values
df['Price'] = df['Price'].replace('Not Priced', np.nan)
df['Price'] = df['Price'].str.replace('$', '').str.replace(',', '').astype(float)

# Numerical columns analysis
numerical_cols = ['Year', 'Price', 'ConsumerRating', 'ConsumerReviews', 'SellerRating', 
                 'SellerReviews', 'ComfortRating', 'InteriorDesignRating', 'PerformanceRating', 
                 'ValueForMoneyRating', 'ExteriorStylingRating', 'ReliabilityRating', 'MinMPG', 
                 'MaxMPG', 'Mileage']

print("Numerical Columns Statistics:\
")
for col in numerical_cols:
    if col in df.columns:
        print(f"\
{col}:")
        stats = df[col].describe()
        print(f"Mean: {stats['mean']:.2f}")
        print(f"Median: {df[col].median():.2f}")
        print(f"Variance: {df[col].var():.2f}")
        print(f"Standard Deviation: {df[col].std():.2f}")
```

```python
print("\
Categorical Columns Analysis:\
")
categorical_cols = ['Make', 'Model', 'Used/New', 'SellerType', 'State', 'DealType', 
                   'ExteriorColor', 'InteriorColor', 'Drivetrain', 'FuelType', 'Transmission']

for col in categorical_cols:
    if col in df.columns:
        print(f"\
{col}:")
        value_counts = df[col].value_counts()
        percentages = (value_counts / len(df) * 100).round(2)
        # Display top 5 categories for each column to avoid overwhelming output
        for val, count in value_counts.head().items():
            print(f"{val}: {count} ({percentages[val]:.1f}%)")
           
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set visualization style
sns.set(style='whitegrid')

# 1. Bar chart for top 10 car makes by frequency
plt.figure(figsize=(10, 6))
make_counts = df['Make'].value_counts().head(10)
sns.barplot(x=make_counts.index, y=make_counts.values, palette='viridis')
plt.title('Top 10 Car Makes')
plt.ylabel('Count')
plt.xlabel('Make')
plt.xticks(rotation=45)
plt.show()
```

```python
# 2. Line graph for number of cars per year
plt.figure(figsize=(10, 6))
year_counts = df['Year'].value_counts().sort_index()
plt.plot(year_counts.index, year_counts.values, marker='o', linestyle='-', color='b')
plt.title('Number of Cars per Year')
plt.ylabel('Count')
plt.xlabel('Year')
plt.grid(True)
plt.show()
```

```python
# Create a zoomed in pie chart with enhanced visibility
plt.figure(figsize=(20, 10))  # Increased figure size
used_new_counts = df['Used/New'].value_counts()
colors = ['#ff9999', '#66b3ff']
plt.pie(used_new_counts.values, labels=used_new_counts.index, autopct='%1.2f%%', 
        startangle=90, colors=colors, 
        textprops={'fontsize': 10},  # Increased font size
        labeldistance=1.1,  # Moved labels slightly outward
        pctdistance=0.8)    # Moved percentage slightly outward
plt.title('Distribution of Used vs New Cars', pad=20, fontsize=20)
plt.axis('equal')
plt.show()
```

```python
# Select only numeric columns
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns
correlation_matrix = df[numeric_columns].corr()

# Create heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap of Numerical Variables', pad=20, fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
